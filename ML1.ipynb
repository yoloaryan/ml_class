# Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

# Sample Data

```python
X=np.array([50,60,760,80,90,100,110,120,130,140]).reshape(-1,1)
y= np.array([150,160,170,180,190,205,210,220,235,240])
```

# Create and train the model

```python
model = LinearRegression()
model.fit(X,y)
```

# Make  predictions

```python
new_spend = np.array([[95]])
predicted_sales = model.predict(new_spend)
```

# Print model parameters

```python
print("Slope : ", model.coef_[0])
print("Intercept : ",model.intercept_)
print("Predicted sales : ", predicted_sales[0])
```

#  Visualize the refgression line

```python
plt.scatter(X,y,color='blue',label='Actual Sales')

plt.plot(X,model.predict(X),color='red',label='Regression Line')

plt.scatter(new_spend,predicted_sales,color="green",s=100,label='Predicted Point')
```
